### Netflix Content Release Trends
Analyze the distribution of Netflix titles by release year.

```python
import pandas as pd

df = pd.read_csv("../../data/processed/netflix_cleaned.csv")
df["release_year"].value_counts().sort_index()
```

### Netflix Historical Release Trends
Visualize Netflix titles by release year using an interactive line chart.

```python
import plotly.express as px

year_counts = df["release_year"].value_counts().sort_index().reset_index()
year_counts.columns = ["Release Year", "Titles"]

fig = px.line(
    year_counts, x="Release Year", y="Titles",
    markers=True,
    color_discrete_sequence=px.colors.sequential.RdBu,
    template="plotly_dark",
    height=500, width=850,
    title="Netflix Historical Release Trends"
)

fig.update_traces(line=dict(width=3), marker=dict(size=7))
fig.write_html("../outputs/release_year_trend.html")
fig.show()
```

### Top 10 Years by Netflix Content Releases
Identify the years with the highest number of titles.

```python
top_years = year_counts.nlargest(10, "Titles").sort_values("Titles")

fig = px.bar(
    top_years, x="Release Year", y="Titles",
    color="Titles",
    color_continuous_scale=px.colors.sequential.RdBu,
    template="plotly_dark",
    title="Top 10 Release Years",
    height=450, width=750
)

fig.write_html("../outputs/top_release_years.html")
fig.show()
```

```python
df["Decade"] = (df["release_year"] // 10) * 10

decade_counts = df["Decade"].value_counts().sort_index().reset_index()
decade_counts.columns = ["Decade", "Titles"]

fig = px.bar(
    decade_counts, x="Decade", y="Titles",
    color="Titles",
    color_continuous_scale=px.colors.sequential.RdBu,
    template="plotly_dark",
    title="Netflix Content by Decade",
    height=450, width=750
)

fig.write_html("../outputs/content_by_decade.html")
fig.show()
```

### Movies vs TV Shows by Release Year
Compare the historical release-year trends of Movies and TV Shows.

```python
year_type = df.groupby(["release_year", "type"]).size().reset_index(name="Titles")

fig = px.line(
    year_type, x="release_year", y="Titles", color="type",
    color_discrete_sequence=px.colors.sequential.RdBu,
    template="plotly_dark", height=500, width=850,
    title="Movies vs TV Shows by Release Year"
)

fig.write_html("../outputs/release_trends_by_type.html")
fig.show()
```

```python
fig = px.area(
    year_counts, x="Release Year", y="Titles",
    color_discrete_sequence=px.colors.sequential.RdBu,
    template="plotly_dark",
    title="Netflix Yearly Release Distribution",
    height=450, width=800
)

fig.write_html("../outputs/yearly_release_area.html")
fig.show()
```

### Release Year Insights
Identify the year with the highest number of Netflix titles.

```python
peak_year = year_counts.loc[year_counts["Titles"].idxmax()]

print("Peak Release Year:", peak_year["Release Year"])
print("Number of Titles:", peak_year["Titles"])
```

### Yearly Content Growth Analysis
Analyze year-over-year changes in Netflix titles by release year.

```python
year_counts["Yearly Change"] = year_counts["Titles"].diff()

fig = px.bar(
    year_counts, x="Release Year", y="Yearly Change",
    color="Yearly Change",
    color_continuous_scale=px.colors.sequential.RdBu,
    template="plotly_dark",
    title="Yearly Change in Netflix Releases",
    height=450, width=800
)

fig.write_html("../outputs/yearly_content_growth.html")
fig.show()
```
